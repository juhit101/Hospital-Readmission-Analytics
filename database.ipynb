```python
# import libraries
import sqlite3 
import pandas as pd
```

```python
# read in dataframe
df_final = pd.read_csv("final_data.csv", index_col=0)
```

```python
df_final.columns
```

```python
# connect to database
conn = sqlite3.connect("final_data.db")
cursor = conn.cursor()
```

```python
# convert df to table
df_final.to_sql('admissions', conn, if_exists='replace', index=False)
```

```python
# check df was properly converted
pd.read_sql("SELECT * FROM admissions", conn)
```

```python
# check data types
cursor.execute("PRAGMA table_info('admissions')")
cursor.fetchall()
```

```python
# read in icd code dataframe
icd_df = pd.read_csv('icd_code.csv')
```

```python
# convert icd code dataframe into table
icd_df.to_sql('icd_codes', conn, if_exists='replace', index=False)
conn.commit()
```

```python
# create table for prediction results
cursor.execute("DROP TABLE IF EXISTS predictions")
cursor.execute("CREATE TABLE predictions AS SELECT * FROM admissions")
cursor.execute("DELETE FROM predictions")
cursor.execute("ALTER TABLE predictions ADD COLUMN readmission_risk FLOAT")
conn.commit()
```
